```python
# Importamos librerías
import numpy as np
import pandas as pd
import warnings
```

```python
# Declaramos nuestras rutas con las bases de datos respectivas
bd_2020 = pd.read_csv('C:/Users/Joyssie/Documents/Data Minning/Proyecto/BDS/2020/BD_FINAL.csv',sep=',')
bd_2021 = pd.read_csv('C:/Users/Joyssie/Documents/Data Minning/Proyecto/BDS/2021/BD_FINAL.csv',sep=',')
bd_2022 = pd.read_csv('C:/Users/Joyssie/Documents/Data Minning/Proyecto/BDS/2022/BD_FINAL.csv', sep=',')
bd_2023 = pd.read_csv('C:/Users/Joyssie/Documents/Data Minning/Proyecto/BDS/2023/BD_FINAL.csv', sep=',')
bd_2024 = pd.read_csv('C:/Users/Joyssie/Documents/Data Minning/Proyecto/BDS/2024/BD_FINAL.csv', sep=',')
bd_2025 = pd.read_csv('C:/Users/Joyssie/Documents/Data Minning/Proyecto/BDS/2025/BD_FINAL.csv', sep=',')
```

```python
# Las hacemos dataframes
df_2020 = pd.DataFrame(bd_2020)
df_2021 = pd.DataFrame(bd_2021)
df_2022 = pd.DataFrame(bd_2022)
df_2023 = pd.DataFrame(bd_2023)
df_2024 = pd.DataFrame(bd_2024)
df_2025 = pd.DataFrame(bd_2025)
```
